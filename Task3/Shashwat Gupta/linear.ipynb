```python
import numpy as np
import pandas as pd

df_insurance = pd.read_csv("insurance.csv")

mapping_dict = {
    "female": 0, "male": 1,
    "no": 0, "yes": 1,
    "northeast": 1, "northwest": 2, "southeast": 3, "southwest": 4
}
df_insurance = df_insurance.replace(mapping_dict)

data_matrix = df_insurance.values
features = data_matrix[:, :6]
target = data_matrix[:, 6]

min_val_x = features.min(axis=0)
max_val_x = features.max(axis=0)
scaled_features = (features - min_val_x) / (max_val_x - min_val_x)

min_val_y = target.min()
max_val_y = target.max()
scaled_target = (target - min_val_y) / (max_val_y - min_val_y)

num_samples, num_features = scaled_features.shape
weights = np.zeros(num_features)
bias = 0.0
learning_rate = 0.1

for epoch in range(1000):
    predictions = np.dot(scaled_features, weights) + bias
    
    grad_w = np.zeros(num_features)
    grad_b = 0.0
    
    for idx in range(num_samples):
        error = predictions[idx] - scaled_target[idx]
        grad_b += error
        for feat_idx in range(num_features):
            grad_w[feat_idx] += error * scaled_features[idx][feat_idx]
            
    weights -= learning_rate * (grad_w / num_samples)
    bias -= learning_rate * (grad_b / num_samples)

print("Optimized Weights:", weights)
print("Optimized Bias:", bias)

sample_input = np.array([19, 0, 27.9, 0, 1, 4]) 

scaled_sample = (sample_input - min_val_x) / (max_val_x - min_val_x)
predicted_scaled_cost = np.dot(scaled_sample, weights) + bias
final_prediction = predicted_scaled_cost * (max_val_y - min_val_y) + min_val_y

print(f"Prediction for provided test is: {final_prediction}")
```
