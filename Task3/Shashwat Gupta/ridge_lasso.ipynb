```python
import pandas as pd
import numpy as np

data_frame = pd.read_csv("Boston.csv")
dataset = data_frame.to_numpy()
dataset

features_matrix = dataset[:, 1:14]
target_vector = dataset[:, 14]
print(features_matrix)
print(target_vector)

min_vals = np.min(features_matrix, axis=0)
max_vals = np.max(features_matrix, axis=0)
features_matrix = (features_matrix - min_vals) / (max_vals - min_vals)

num_samples, num_features = features_matrix.shape
learning_rate = 0.01
reg_factor = 1

weights_ridge = np.zeros(num_features)
bias_ridge = 0.0
weights_lasso = np.zeros(num_features)
bias_lasso = 0.0

for epoch in range(1000):
    predictions = np.dot(features_matrix, weights_ridge) + bias_ridge
    grad_w = np.zeros(num_features)
    grad_b = 0.0
    
    for row_idx in range(num_samples):
        error = predictions[row_idx] - target_vector[row_idx]
        grad_b += error
        for col_idx in range(num_features):
            grad_w[col_idx] += error * features_matrix[row_idx][col_idx]
            
    grad_w = (grad_w / num_samples) + (reg_factor * weights_ridge) / num_samples
    grad_b = grad_b / num_samples
    
    weights_ridge -= learning_rate * grad_w
    bias_ridge -= learning_rate * grad_b

for epoch in range(1000):
    predictions = np.dot(features_matrix, weights_lasso) + bias_lasso
    grad_w = np.zeros(num_features)
    grad_b = 0.0
    
    for row_idx in range(num_samples):
        error = predictions[row_idx] - target_vector[row_idx]
        grad_b += error
        for col_idx in range(num_features):
            grad_w[col_idx] += error * features_matrix[row_idx][col_idx]
            
    grad_w = (grad_w / num_samples) + (reg_factor * np.sign(weights_lasso)) / num_samples
    grad_b = grad_b / num_samples
    
    weights_lasso -= learning_rate * grad_w
    bias_lasso -= learning_rate * grad_b

outputs_ridge = np.dot(features_matrix, weights_ridge) + bias_ridge
outputs_lasso = np.dot(features_matrix, weights_lasso) + bias_lasso

mse_ridge = 0.0
mse_lasso = 0.0
for idx in range(num_samples):
    mse_ridge += (outputs_ridge[idx] - target_vector[idx]) ** 2
    mse_lasso += (outputs_lasso[idx] - target_vector[idx]) ** 2
    
mse_ridge /= num_samples
mse_lasso /= num_samples

print("Ridge MSE =", mse_ridge)
print("Lasso MSE =", mse_lasso)

columns = ["crim", "zn", "indus", "chas", "nox", "rm", "age", "dis", "rad", "tax", "ptratio", "black", "lstat"]
print("RIDGE COEFFICIENTS")
for i in range(num_features):
    print(columns[i], ":", weights_ridge[i])
print()
print("LASSO COEFFICIENTS")
for i in range(num_features):
    print(columns[i], ":", weights_lasso[i])

test_sample = np.array([0.00632, 18.0, 2.31, 0, 0.538, 6.575, 65.2, 4.09, 1, 296, 15.3, 396.9, 4.98])

test_sample = (test_sample - min_vals) / (max_vals - min_vals)

pred_val_ridge = np.dot(test_sample, weights_ridge) + bias_ridge
pred_val_lasso = np.dot(test_sample, weights_lasso) + bias_lasso

print("Ridge Prediction =", pred_val_ridge)
print("Lasso Prediction =", pred_val_lasso)
```
