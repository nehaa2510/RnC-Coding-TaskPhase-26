```python
import numpy as np
import pandas as pd

dataset = pd.read_csv("heart.csv")
data_matrix = dataset.values
data_matrix

features = data_matrix[:, :12]
target = data_matrix[:, 13]
print(features)
print(target)

min_vals = np.min(features, axis=0)
max_vals = np.max(features, axis=0)

features = (features - min_vals) / (max_vals - min_vals)
features

num_samples, num_features = features.shape
weights = np.zeros(num_features)
bias = 0.0
learning_rate = 0.1

def compute_sigmoid(activation_input):
    return 1 / (1 + np.exp(-activation_input))

for epoch in range(1000):
    linear_model = np.dot(features, weights) + bias
    predictions = compute_sigmoid(linear_model)
    
    grad_weights = np.zeros(num_features)
    grad_bias = 0.0
    
    for idx in range(num_samples):
        error = predictions[idx] - target[idx]
        grad_bias += error
        for feat_idx in range(num_features):
            grad_weights[feat_idx] += error * features[idx][feat_idx]
            
    weights -= learning_rate * (grad_weights / num_samples)
    bias -= learning_rate * (grad_bias / num_samples)

linear_model = np.dot(features, weights) + bias
probabilities = compute_sigmoid(linear_model)
class_predictions = np.array([1 if p >= 0.5 else 0 for p in probabilities])

sample_test = np.array([63, 1, 3, 145, 233, 1, 0, 150, 0, 2.3, 0, 0])

sample_test = (sample_test - min_vals) / (max_vals - min_vals)
test_linear = np.dot(sample_test, weights) + bias
test_prob = compute_sigmoid(test_linear)

print("Probability =", test_prob)
if test_prob >= 0.5:
    print("Heart Disease")
else:
    print("No Heart Disease")
```
