```python
import numpy as np
import pandas as pd
db=pd.read_csv("insurance.csv")
db=db.replace({"male":0,"female":1,"yes":1,"no":0,"southwest":1,"southeast":2,"northwest":3,"northeast":0})
db=db.values
db
```

```python
x=db[:,:6]
y=db[:,6]
print(x)
print(y)
```

```python
x_min=x.min(axis=0)
x_max=x.max(axis=0)
x=(x-x_min)/(x_max-x_min)
print(x)
```

```python
y_min=y.min()
y_max=y.max()
y=(y-y_min)/(y_max-y_min)
print(y)
```

```python
m,n=x.shape
w=np.zeros(n)
b=0
a=0.1
```

```python
for k in range(1000):
    f_wb=np.dot(x,w)+b
    for j in range(n):
        dj_dw=0
        dj_db=0
        for  i in range(m):
            dj_dw+=(f_wb[i]-y[i])*x[i][j]
            dj_db+=f_wb[i]-y[i]
        dj_dw=dj_dw/m
        dj_db=dj_db/m
        w[j]-=a*dj_dw
    b-=a*dj_db
    
```

```python
print(w)
b
```

```python
test=[19,1,27.9,0,1,1]
```

```python
test=(test-x_min) / (x_max - x_min)
insuranceprediction=np.dot(test,w)+b
insuranceprediction=insuranceprediction*(y_max-y_min)+y_min
print("Prediction for provided test is:",insuranceprediction)
```
