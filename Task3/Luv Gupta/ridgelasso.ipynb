```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("Boston.csv")
df=df.values
df
```

```python
x=df[:,1:14]
y=df[:,14]
print(x)
print(y)
```

```python
x_min=np.min(x,axis=0)
x_max=np.max(x,axis=0)
x=(x-x_min)/(x_max-x_min)
```

```python
m,n=x.shape
a=0.01
lambda_reg=1
w_ridge=np.zeros(n)
b_ridge=0
w_lasso=np.zeros(n)
b_lasso=0
```

```python
for i in range(1000):
    prediction=np.dot(x,w_ridge)+b_ridge
    dj_dw=np.zeros(n)
    dj_db=0
    for j in range(m):
        dj_db+=(prediction[j]-y[j])
        for k in range(n):
            dj_dw[k]+=(prediction[j]-y[j])*x[j][k]
    dj_dw=(dj_dw/m)+(lambda_reg*w_ridge)/m
    dj_db=dj_db/m
    w_ridge-=a*dj_dw
    b_ridge-=a*dj_db
```

```python
for i in range(1000):
    y_pred=np.dot(x,w_lasso)+b_lasso
    dj_dw=np.zeros(n)
    dj_db=0
    for j in range(m):
        dj_db+=(y_pred[j]-y[j])
        for k in range(n):
            dj_dw[k]+=(y_pred[j]-y[j])*x[j][k]
    dj_dw=(dj_dw/m)+(lambda_reg*np.sign(w_lasso))/m
    dj_db=dj_db/m
    w_lasso-=a*dj_dw
    b_lasso-=a*dj_db
```

```python
ridge_pred=np.dot(x,w_ridge)+b_ridge
lasso_pred=np.dot(x,w_lasso)+b_lasso
```

```python
ridge_mse=0
lasso_mse=0
for i in range(m):
    ridge_mse+=(ridge_pred[i]-y[i])**2
    lasso_mse+=(lasso_pred[i]-y[i])**2
ridge_mse=ridge_mse/m
lasso_mse=lasso_mse/m
print("Ridge MSE =",ridge_mse)
print("Lasso MSE =",lasso_mse)
#we can see lasso is better
```
