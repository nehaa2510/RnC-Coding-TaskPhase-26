```python
import numpy as np
import pandas as pd
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
df=pd.read_csv("diabetes.csv")
df=df.values
df
```

```python
x=df[:,:8]
y=df[:,8]
print(x)
y
```

```python
for i in range(len(y)):
    if y[i]==0:
        y[i]=-1
```

```python
x_min=np.min(x,axis=0)
x_max=np.max(x,axis=0)
x=(x-x_min)/(x_max-x_min)
x
```

```python
m,n=x.shape
w=np.zeros(n)
b=0
a=0.001
lambda_reg=0.01
```

```python
for k in range(1000):
    for i in range(m):
        f_wb=np.dot(x[i],w)+b
        if y[i]*f_wb>=1:
            for j in range(n):
                w[j]=w[j]-a*(2*lambda_reg*w[j])
        else:
            for j in range(n):
                w[j]=w[j]-a*(2*lambda_reg*w[j]-y[i]*x[i][j])
            b=b+a*y[i]
```

```python
prediction=np.zeros(m)

for i in range(m):

    f_wb=np.dot(x[i],w)+b

    if f_wb>=0:
        prediction[i]=1
    else:
        prediction[i]=-1
```

```python
correct=0
for i in range(m):
    if prediction[i]==y[i]:
        correct+=1
accuracy=(correct/m)*100
print("Accuracy =",accuracy)
```
