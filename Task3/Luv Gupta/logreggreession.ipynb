```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("heart.csv")
df=df.values
df
```

```python
x=df[:,:12]      
y=df[:,13]
print(x)
print(y)      
```

```python
x_min=np.min(x,axis=0)
x_max=np.max(x,axis=0)

x=(x-x_min)/(x_max-x_min)
x
```

```python
m,n=x.shape
w=np.zeros(n)
b=0
a=0.1
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
for k in range(1000):
    z=np.dot(x,w)+b
    f_wb=sigmoid(z)
    dj_db=0
    for j in range(n):
        dj_dw=0
        for i in range(m):
            dj_dw+=(f_wb[i]-y[i])*x[i][j]
            if j == 0:
                dj_db+=(f_wb[i]-y[i])
        dj_dw=dj_dw/m
        w[j]=w[j]-a*dj_dw
    dj_db=dj_db/m
    b=b-a*dj_db
```

```python
z=np.dot(x,w)+b
prob=sigmoid(z)
prediction=np.zeros(m)
for i in range(m):
    if prob[i]>=0.5:
        prediction[i]=1
    else:
        prediction[i]=0
```

```python
test=np.array([63,1,3,145,233,1,0,150,0,2.3,0,0])
```

```python
test=(test-x_min)/(x_max-x_min)
z=np.dot(test,w)+b
prob=sigmoid(z)
print("Probability =",prob)
if prob>=0.5:
    print("Heart Disease")
else:
    print("No Heart Disease")
```
