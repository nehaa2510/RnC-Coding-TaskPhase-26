```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Boston.csv")

print(df.head())
```

```python
X = df.drop(["medv","Unnamed: 0"], axis = 1).values
y = df["medv"].values

print(X)
print(y)
```

```python
X_max = X.max(axis=0)
X_min = X.min(axis=0)
y_max = y.max()
y_min = y.min()

def normalise(data):
    data_norm = (data - data.min(axis=0))/(data.max(axis=0)-data.min(axis=0))
    return data_norm

```

```python
X_norm = normalise(X)
print(X_norm)
```

```python
m,n = X_norm.shape
a = 0.1
itr = 2000
lambda_reg = 1
```

```python
#Ridge model


w_ridge = np.zeros(n)
b_ridge = 0
```

```python
for i in range(itr):
    y_pred = np.dot(X_norm , w_ridge) + b_ridge
   

    error = y_pred - y


    dj_dw = (np.dot(X_norm.T, error) / m) + (lambda_reg * w_ridge) / m
    dj_db = np.sum(error) / m

    w_ridge -= a * dj_dw
    b_ridge -= a * dj_db


print("ridge w: ", w_ridge)
print("ridge b: ",b_ridge)
```

```python
#lasso model


w_lasso = np.zeros(n)
b_lasso = 0

for i in range(itr):
    y_pred = np.dot(X_norm, w_lasso) + b_lasso

   
    error = y_pred - y


    dj_dw = (np.dot(X_norm.T, error) / m) + (lambda_reg * np.sign(w_lasso)) / m
    dj_db = np.sum(error) / m

    w_lasso -= a * dj_dw
    b_lasso -= a * dj_db
    

print("lasso w: ", w_lasso)
print("lasso b: ", b_lasso )    
```

```python
#comparision


ridge_mse = np.mean((np.dot(X_norm , w_ridge)+ b_ridge - y)**2)
lasso_mse = np.mean((np.dot(X_norm,w_lasso) +b_lasso - y)**2)


print("ridge mse: ", ridge_mse)
print("Lasso mse: ", lasso_mse)
```
