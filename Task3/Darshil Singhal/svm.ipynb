```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("diabetes.csv")
```

```python
X = df.drop(["Outcome"], axis = 1).values
y = df["Outcome"].values


y = np.where(y==0,-1,1)

```

```python
def normalise(data):
    data_norm = (data - data.min(axis=0)) / (data.max(axis=0) - data.min(axis=0))
    return data_norm

X_norm = normalise(X)
print(X_norm)
```

```python
m,n = X_norm.shape
a = 0.1
itr = 5000
lambda_reg = 0.01

w = np.zeros(n)
b = 0


for i in range(itr):

    dj_dw = lambda_reg*w
    dj_db = 0

    condition = y * (np.dot(X_norm, w) + b)


    for j in range(m):  
        if condition[j] < 1:
            dj_db -= y[j]
            
            dj_dw -= y[j] * X_norm[j]


    dj_dw = dj_dw/m
    dj_db = dj_db/m 

    w -= a*dj_dw
    b-= a*dj_db
```

```python
print("SVM w: ", w)
print("SVM b: ", b)
```

```python
y_hat = np.dot(X_norm, w) + b

y_final = np.sign(y_hat)


tp =0
fp=0
fn=0
tn=0


for j in range(m):
    if y_final[j] == 1 and y[j] ==1:
        tp+=1
    elif y_final[j] == 1 and y[j] == -1:
        fp+=1
    elif y_final[j] == -1 and y[j] == 1:
        fn+=1
    elif y_final[j] == -1 and y[j] == -1:
        tn+=1            



print(f"TP: {tp}, FP: {fp}, TN: {tn}, FN: {fn}")        
```

```python
accuracy = (tp+tn)/m*100
precision = tp/(tp+fp) if(tp+fp) > 0 else 0
recall = tp/ (tp+fn) if (tp+fn ) > 0 else 0
f1_score = 2*(precision *recall) / (precision+recall) if (precision + recall) > 0 else 0


```

```python

print("Accuracy: ", accuracy)
print("Precision: ", precision)
print("Recall: ", recall)
print("F1 Score: ", f1_score)

```
